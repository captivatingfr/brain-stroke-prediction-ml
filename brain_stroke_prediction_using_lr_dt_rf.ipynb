Transforming categorical column using encoding

split the dataset using train test split

```python
from imblearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
from sklearn.preprocessing import StandardScaler
```

```python
X = df.drop(['stroke'], axis=1)
y = df['stroke']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, stratify=y, random_state=42)
```

```python
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.pipeline import Pipeline

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
```

Logistic Regression

```python
lr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression(class_weight='balanced', random_state=42, max_iter=1000))
])

lr_param_grid = {
    'classifier__C': [0.001, 0.01, 0.1, 1, 10] # Regularization strength
}

lr_grid = GridSearchCV(lr_pipeline, param_grid=lr_param_grid, cv=skf, scoring='recall', n_jobs=-1)
lr_grid.fit(X_train, y_train)

lr_pred = lr_grid.predict(X_test)
lr_prob = lr_grid.predict_proba(X_test)[:, 1]

print("Best LR Parameters:", lr_grid.best_params_)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm=confusion_matrix(y_test, lr_pred)
```

```python
print(cm)
```

```python
from sklearn.metrics import classification_report, roc_auc_score

lr_prob = lr_grid.predict_proba(X_test)[:, 1]

print("--- Logistic Regression Performance ---")
print(classification_report(y_test, lr_pred, target_names=['No Stroke', 'Stroke']))
print(f"ROC-AUC Score: {roc_auc_score(y_test, lr_prob):.4f}")
```

```python
from sklearn import metrics
```

```python
sns.heatmap(confusion_matrix(y_test, lr_pred), annot=True, fmt='d')
plt.xlabel("Prediction")
plt.ylabel("Test")
plt.show()
```

Decision Tree

```python
dt_pipeline = Pipeline([
    ('classifier', DecisionTreeClassifier(class_weight='balanced', random_state=42))
])

dt_param_grid = {
    'classifier__max_depth': [3, 5, 7, 10],
    'classifier__min_samples_split': [2, 5, 10],
    'classifier__criterion': ['gini', 'entropy']
}

dt_grid = GridSearchCV(dt_pipeline, param_grid=dt_param_grid, cv=skf, scoring='recall', n_jobs=-1)
dt_grid.fit(X_train, y_train)

dt_pred = dt_grid.predict(X_test)
dt_prob = dt_grid.predict_proba(X_test)[:, 1]

print("Best DT Parameters:", dt_grid.best_params_)
```

```python
cm=confusion_matrix(y_test,dt_pred)
cm
```

```python
from sklearn.metrics import confusion_matrix

sns.heatmap(confusion_matrix(y_test, dt_pred), annot=True, fmt='d')
plt.xlabel("Prediction")
plt.ylabel("Test")
plt.show()
```

Random Forest

```python
rf_pipeline = Pipeline([
    ('classifier', RandomForestClassifier(class_weight='balanced', random_state=42))
])

rf_param_grid = {
    'classifier__n_estimators': [50, 100, 200],
    'classifier__max_depth': [5, 10, 15],
    'classifier__min_samples_leaf': [1, 2, 4]
}

rf_grid = GridSearchCV(rf_pipeline, param_grid=rf_param_grid, cv=skf, scoring='recall', n_jobs=-1)
rf_grid.fit(X_train, y_train)

rf_pred = rf_grid.predict(X_test)
rf_prob = rf_grid.predict_proba(X_test)[:, 1]

print("Best RF Parameters:", rf_grid.best_params_)
```

```python
cm = confusion_matrix(y_test, rf_pred)
cm
```

```python
rf_pred = rf_grid.predict(X_test)
rf_prob = rf_grid.predict_proba(X_test)[:, 1]
custom_threshold = 0.25
rf_pred_tuned = (rf_prob >= custom_threshold).astype(int)

print("--- Random Forest Performance ---")
print(classification_report(y_test, rf_pred_tuned, target_names=['No Stroke', 'Stroke']))
print(f"ROC-AUC Score: {roc_auc_score(y_test, rf_prob):.4f}")
```

```python
from sklearn.metrics import recall_score, roc_auc_score, precision_score, f1_score

sns.heatmap(confusion_matrix(y_test, rf_pred_tuned), annot=True, fmt='d')
plt.xlabel("Prediction")
plt.ylabel("Test")
plt.show()
```

```python
models = {
    'Logistic Regression': (lr_pred, lr_prob),
    'Decision Tree': (dt_pred, dt_prob),
    'Random Forest': (rf_pred, rf_prob)
}
comparison_data = []

for name, (preds, probs) in models.items():
    comparison_data.append({
        'Model': name,
        'ROC-AUC': roc_auc_score(y_test, probs),
        'Stroke Recall': recall_score(y_test, preds, pos_label=1),
        'Stroke Precision': precision_score(y_test, preds, pos_label=1, zero_division=0),
        'Stroke F1-Score': f1_score(y_test, preds, pos_label=1)
    })
```

```python

import os

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

os.makedirs("assets", exist_ok=True)

# ---------------------------------------------------------------------
# ---------------------------------------------------------------------
trained_models = {
    "Logistic Regression": lr_grid,     # <- updated to use your fitted Logistic Regression grid search
    "Decision Tree": dt_grid,           # <- updated to use your fitted Decision Tree grid search
    "Random Forest": rf_grid,           # <- updated to use your fitted Random Forest grid search
}
RF_NAME = "Random Forest"
DF = df  # <- your dataframe that has a "stroke" column and an "age" column

# X_train, X_test, y_train, y_test must be your 80:20 split variables.
# NOTE: if one of your models was trained on SCALED features, evaluate it
# on the matching scaled test set instead of X_test.

# ---------------------------------------------------------------------
# 2. Add KNN and Naive Bayes (new, trained here on the same split)
# ---------------------------------------------------------------------
knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
knn.fit(X_train, y_train)
nb = GaussianNB().fit(X_train, y_train)
trained_models["KNN"] = knn
trained_models["Naive Bayes"] = nb

y_true = np.asarray(y_test)

# ---------------------------------------------------------------------
# 3. Comparison table (default 0.5 threshold for every model)
# ---------------------------------------------------------------------
row = []
rows = []
for name, m in trained_models.items():
    proba = m.predict_proba(X_test)[:, 1]
    pred = (proba >= 0.5).astype(int)
    rows.append({
        "Model": name,
        "Accuracy": round(accuracy_score(y_true, pred), 4),
        "Precision": round(precision_score(y_true, pred, zero_division=0), 4),
        "Recall": round(recall_score(y_true, pred, zero_division=0), 4),
        "F1": round(f1_score(y_true, pred, zero_division=0), 4),
        "ROC-AUC": round(roc_auc_score(y_true, proba), 4),
    })
metrics_df = pd.DataFrame(rows)
metrics_df.to_csv("assets/metrics_comparison.csv", index=False)
print(metrics_df)

# ---------------------------------------------------------------------
# 4. Random Forest threshold-tuning table
# ---------------------------------------------------------------------
rf_proba = trained_models[RF_NAME].predict_proba(X_test)[:, 1]
thr_rows = []
for t in np.arange(0.05, 0.96, 0.05):
    pred = (rf_proba >= t).astype(int)
    thr_rows.append({
        "Threshold": round(float(t), 2),
        "Precision": round(precision_score(y_true, pred, zero_division=0), 4),
        "Recall": round(recall_score(y_true, pred, zero_division=0), 4),
        "F1": round(f1_score(y_true, pred, zero_division=0), 4),
        "False Positives": int(((pred == 1) & (y_true == 0)).sum()),
        "False Negatives": int(((pred == 0) & (y_true == 1)).sum()),
    })
pd.DataFrame(thr_rows).to_csv("assets/threshold_results.csv", index=False)

# ---------------------------------------------------------------------
# 5. Images for the EDA / evaluation tab
# ---------------------------------------------------------------------
counts = DF["stroke"].value_counts().sort_index()
plt.figure(figsize=(5, 4))
plt.bar(["No stroke (0)", "Stroke (1)"], counts.values, color=["#0f766e", "#b45309"])
plt.ylabel("Records")
plt.title("Class distribution")
plt.tight_layout()
plt.savefig("assets/class_distribution.png", dpi=150)
plt.close()

plt.figure(figsize=(6, 4))
plt.hist(DF.loc[DF["stroke"] == 0, "age"], bins=20, alpha=0.6, label="No stroke (0)", color="#0f766e")
plt.hist(DF.loc[DF["stroke"] == 1, "age"], bins=20, alpha=0.8, label="Stroke (1)", color="#b45309")
plt.xlabel("Age")
plt.ylabel("Records")
plt.title("Age distribution by class")
plt.legend()
plt.tight_layout()
plt.savefig("assets/age_by_stroke.png", dpi=150)
plt.close()

n = len(trained_models)
fig, axes = plt.subplots(1, n, figsize=(4 * n, 4))
for ax, (name, m) in zip(np.atleast_1d(axes), trained_models.items()):
    pred = (m.predict_proba(X_test)[:, 1] >= 0.5).astype(int)
    ConfusionMatrixDisplay.from_predictions(y_true, pred, ax=ax, colorbar=False)
    ax.set_title(name)
plt.tight_layout()
plt.savefig("assets/confusion_matrices.png", dpi=150)
plt.close()

plt.figure(figsize=(6, 5))
for name, m in trained_models.items():
    fpr, tpr, _ = roc_curve(y_true, m.predict_proba(X_test)[:, 1])
    plt.plot(fpr, tpr, label=name)
plt.plot([0, 1], [0, 1], "k--", alpha=0.5)
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curves")
plt.legend()
plt.tight_layout()
plt.savefig("assets/roc_curves.png", dpi=150)
plt.close()

print("Done. Files in assets/:", os.listdir("assets"))

# ---------------------------------------------------------------------
# 6. Colab only: zip the folder and download it
# ---------------------------------------------------------------------
try:
    import shutil
    from google.colab import files

    shutil.make_archive("assets", "zip", "assets")
    files.download("assets.zip")
except Exception:
    pass
```
